```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
companies = pd.read_csv("Unicorn_Companies.csv")
companies.head()
```

```python
companies.info()
companies.shape
```

```python
companies.drop_duplicates().shape
```

```python
companies_sorted=companies.sort_values('Year Founded',ascending=False)
companies_sorted.head(10)
```

```python
companies_sorted['Year Founded'].value_counts()
```

```python
sns.histplot(data=companies,x='Year Founded')
plt.title('Year Founded Histogram')
```

```python
companies.info()
```

```python
companies['Date Joined']=pd.to_datetime(companies['Date Joined'])
companies.info()
```

```python
companies['Month Joined']=companies['Date Joined'].dt.month
companies.head(10)
```

```python
companies['Years Joined']=companies['Date Joined'].dt.year
companies.head(10)
```

```python
companies['Years Needed to join']=companies['Years Joined']-companies['Year Founded']
companies.head()
```

```python
companies_2020=companies[companies['Year Founded']==2020]
companies_2020.head(10)
```

```python
companies_2020['Years Needed to join'].value_counts()
```

```python
companies_2020['week joined']=companies_2020['Date Joined'].dt.isocalendar().week
companies_2020.head(10)
```

```python
companies_by_week_2020=companies_2020[['Company','week joined']].groupby(['week joined']).count().reset_index().rename(columns={'Company':'Companies Count'})
companies_by_week_2020
```

```python
companies['month_name']=companies['Date Joined'].dt.month_name()
companies.head()
```

```python
companies1=companies[['month_name','Years Needed to join']]
companies1
```

```python
month_order=["January", "February", "March", "April", "May", "June","July", "August", "September", "October", "November", "December"]
month_order
```

```python
sns.boxplot(data=companies,x='Years Needed to join',y='month_name',order=month_order,showfliers=False)
plt.title('Years needed to join unicorn companies based on months')
plt.show()
```

```python
plt.figure(figsize = (10, 6))
sns.barplot(data=companies,x='Year Founded',y='Years Needed to join',ci=False)
plt.xticks(rotation=45,horizontalalignment='right')
plt.show()
```

```python
plt.figure(figsize=(20,5))
plt.bar(companies_by_week_2020['week joined'],companies_by_week_2020['Companies Count'])
plt.plot()
plt.xlabel('Weeks of 2020')
plt.ylabel('Count of companies joined in those weeks')
plt.title('Number of companies that became unicorns per week in 2020')
plt.xticks(rotation=45,horizontalalignment='right',fontsize=10)
plt.show()
```
